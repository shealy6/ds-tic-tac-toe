```python
#Defining a function for move validation
def check_move(move_1: int, board_1):
    if move_1 not in [1, 2, 3, 4, 5, 6, 7, 8, 9]:
        print("Choose a valid move")
        return False
    elif board_1[move_1-1] == ("X", "O"):
        print("Choose an open space")
        return False
    else:
        print("Nice move")
        return True

check_move(1, board)
```

```python
#Full function for moves

#show board first
show_board(board)

#Define a move number
#move_number = 1

#ask for input and check move is valid, return a valid move
def get_move(board):
    while True:
        try:
            move = int(input("Your move (1-9): "))
        except ValueError:
            print("Please enter a number.")  #Deals with errors for non-integer values
            continue
        if check_move(move, board): # Checks moves validity, if false, prints why and returns to start of the loop
            return move

# Know whose turn it is
for move_number in range(1, 10):
    player = "X" if move_number % 2 == 1 else "O"
    print(f"{player}'s turn. Choose a number to make your move!")
    if move_number == 9:
        print("It's a draw! Try again?")

show_board(board)
#move_number += 1

# Checking for a win after a player moves
def check_winner(board, player):
    rows = [board[0:3], board[3:6], board[6:9]]     # defining three in a row
    cols = [board[0::3], board[1::3], board[2::3]]  # defining three in a column
    diags = [board[0::4], board[2:7:2]]             # defining three in a diagonal
    win = [player] * 3                              # making three Xs or Os depending on the turn
    return win in rows + cols + diags               # Returns True if it finds three in a row

#return a draw after 9 moves
# automatically covered in the turn tracker

 
```
